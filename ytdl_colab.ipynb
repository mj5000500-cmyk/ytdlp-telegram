# 🚀 **ytdl-telegram: YouTube to Telegram Bot on Google Colab**

High-speed YouTube downloader and uploader bot running entirely on Google Colab's cloud infrastructure (1 Gbps+ line speed).

### ✨ Features:
* **Send any YouTube link directly to the bot**: Fetches video title, channel, duration, and thumbnail preview.
* **Quality selection buttons with MB sizes**: Choose 1080p, 720p, 480p, 360p, or MP3 Audio with estimated file sizes.
* **Live Download & Upload Speed**: Real-time progress bar, speed (MB/s), and ETA.
* **In-Chat Cookies Management**: Upload `cookies.txt` directly in Telegram chat or manage via `/cookies`.
* **FastTelethon Parallel Uploads**: Pipelined streaming saturates cloud upload speeds (5-30+ MB/s).
* **Batch Playlist Support**: Sequential processing (`/playlist`) so disk never overflows.
